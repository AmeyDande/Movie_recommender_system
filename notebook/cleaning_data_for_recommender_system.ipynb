```python


def safe_parse(x):
    try:
        return ast.literal_eval(x)
    except (ValueError, SyntaxError):
        return np.nan

crew['crew'] = crew['crew'].apply(safe_parse)
```

```python
crew.iloc[0]
```

```python
crew['crew_info'] = crew['crew'].apply(lambda x: next((i['name'] for i in x if i['department'] == 'Directing'), np.nan) if isinstance(x, list) else np.nan)
```
